# CAS number

```python
df_CAS_No = df['CAS No']
df_CAS_No
```

```python
labels, counts = np.unique(df_CAS_No, return_counts=True)
plt.bar(labels, counts, align='center')
plt.title('Registres CAS')
plt.xlabel('CAS')
plt.ylabel('Nº de registres')
plt.gca().set_xticks(labels)
plt.rcParams["figure.figsize"] = (20,5)
plt.show()
```

```python
# Ruta y nombre de archivo de salida
output_path = '../promisces/promisces-backend/src/main/resources/data/data/first information/CAS_no.xlsx'

# Guardar el DataFrame en formato XLSX
df_CAS_No.to_excel(output_path, index=False)
```

# P-score UBA_W

```python
for i in df['P-score UBA_W (based on equation 2 in https://www.researchgate.net/publication/365285280_Screening_and_prioritising_Persistent_Mobile_and_Toxic_chemicals_development_and_application_of_a_robust_scoring_system)']:
    print(i)
```

```python
# Obtén los datos de los compuestos y sus puntajes
compuestos = df['CAS No']
scores =  df['P-score UBA_W (based on equation 2 in https://www.researchgate.net/publication/365285280_Screening_and_prioritising_Persistent_Mobile_and_Toxic_chemicals_development_and_application_of_a_robust_scoring_system)']

# Configura el tamaño de la figura
plt.figure(figsize=(20, 6))

# Crea el gráfico de barras
plt.bar(compuestos, scores)

# Configura las etiquetas y título del gráfico
plt.xlabel('Compuesto')
plt.ylabel('Score')
plt.title('Puntajes de Compuestos Químicos')

# Ajusta la orientación de las etiquetas en el eje x para evitar superposiciones
plt.xticks(rotation=90)

# Muestra el gráfico
plt.show()
```
