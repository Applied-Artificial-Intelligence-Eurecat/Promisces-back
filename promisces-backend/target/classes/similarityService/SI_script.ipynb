```python
test = df['SMILES'][:20]
```

```python
# Load dataset and compute fingerprints
dataset_smiles = test  # List of SMILES strings for your 120k compounds
dataset_fingerprints = [AllChem.GetMorganFingerprintAsBitVect(Chem.MolFromSmiles(smiles), 2, nBits=2048) for smiles in dataset_smiles]
```

```python
def compute_similarity(fp_query, fp_database):
    return DataStructs.TanimotoSimilarity(fp_query, fp_database)
```

```python
# query_smiles = "C1CC2=C(C=C1)C(=O)CC3C2CCC4C3(CCC4O)C"  # Example estrogen SMILES for query molecule

query_smiles = "C[C@@H]1C2=CC(=O)CC[C@@]2(C)[C@H]3CCc4c(cccc4O3)O1" # estradiol
query_molecule = Chem.MolFromSmiles(query_smiles)
query_fp = AllChem.GetMorganFingerprintAsBitVect(query_molecule, 2, nBits=2048)
```

```python
similarities = []
for i, fp in enumerate(dataset_fingerprints):
    similarity_score = compute_similarity(query_fp, fp)
    similarities.append((i, similarity_score))  # Store index and similarity score

# Sort by similarity score in descending order and retrieve top 5
top_5 = sorted(similarities, key=lambda x: x[1], reverse=True)[:5]
```

```python
for idx, score in top_5:
    print(f"Compound {idx} - Similarity Score: {score}")
    print(f"SMILES: {dataset_smiles[idx]}")
```
